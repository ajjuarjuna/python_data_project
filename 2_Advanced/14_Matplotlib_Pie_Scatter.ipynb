```python
# impoting libraries 
import pandas as pd
import matplotlib.pyplot as plt 
from datasets import load_dataset
import ast

# loading dataset 
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup 
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df['job_work_from_home'].value_counts().plot(kind='pie', startangle= 90, autopct = '%.1f%%')
plt.title("Percentage of Jobs Work from Home")
```

```python
dict1 = {
'job_work_from_home':'Work From Home',
'job_no_degree_mention':'No Job Degree Req.',
'job_health_insurance':'Health Insurance Offered'
}

fig, ax = plt.subplots(1,3)

for i, (col, title) in enumerate(dict1.items()):
    df[col].value_counts().plot(kind='pie',autopct ='%.1f%%', ax=ax[i], title = title, startangle=90)

plt.tight_layout()
```

```python
# with matplotlib plot
fig, ax = plt.subplots(1,3)

for i, (column, title) in enumerate(dict1.items()):
    ax[i].pie(df[column].value_counts(), startangle=90, autopct = '%.1f%%', labels=['False','True'])
    ax[i].set_title(title)

fig.tight_layout()
plt.show()
```
