```python
# importing libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt


# loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()


# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.pivot_table(values = 'salary_year_avg',columns = 'job_title_short', index = 'job_country', aggfunc = 'median')
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.groupby('job_country').size().sort_values(ascending = False)
```

```python
top_countries = df["job_country"].value_counts().head(6).index
top_countries
```

```python
df_job_country_salary = df.pivot_table(
    index = 'job_country',
    columns = 'job_title_short', 
    values = 'salary_year_avg',
    aggfunc = 'median')
    

df_job_country_salary = df_job_country_salary.loc[top_countries]

job_title = ['Data Analyst','Data Engineer', 'Data Scientist']

df_job_country_salary = df_job_country_salary[job_title]

df_job_country_salary.plot(kind = 'bar')
plt.ylabel('Median Salary ( USD $ )')
plt.xlabel('')
plt.xticks(rotation =  45)
plt.title('Median Salary of country and job tiles')
plt.show()
```
