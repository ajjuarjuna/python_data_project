```python
# importing libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt


# loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()


# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df['Month'] = df['job_posted_date'].dt.strftime('%b')
months = df['Month'].unique()
dfdict = {month : df[df['Month'] == month] for month in months}
dfdict['Jan']
df_q1 = pd.concat([dfdict["Jan"],dfdict["Feb"],dfdict["Mar"]],ignore_index = True)
```

## To Csv

```python
df_q1.to_csv("quarter1.csv")
pd.read_csv('quarter1.csv', index_col = 0)
```

## To Excel

```python
df_q1.to_excel('qaurter1_.xlsx')
```

## To Sql

```python
from sqlalchemy import create_engine
engine = create_engine('sqllite:///jobs.db')
df.to_sql('job_table', con=engine, if_exists='append', index=False)
```

## To parquet

```python
df_q1.to_parquet('job_data.parquet')
```

## To Pickle

```python
df_q1.to_pickle('job_data.pkl')
```
