```python
# importing libraries 
import pandas as pd 
import matplotlib.pyplot as plt
from datasets import load_dataset

# loading dataset 
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])].copy()
```

```python
def calculate_salary(salary):
    return salary * 1.03
```

```python
df_salary['inflated_salary'] = df['salary_year_avg'].apply(calculate_salary)
```

```python
df_salary[['salary_year_avg','inflated_salary']]
```

```python
df_salary['inflated_salary2'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)
```

```python
df_salary[['salary_year_avg','inflated_salary','inflated_salary2']]
```

## Apply function with rows

Senior 1.05  
Rest 1.03

```python
def project_salary(row):
    if 'Senior' in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else :
        return 1.03 * row['salary_year_avg']
```

```python
df_salary['inflated_salary'] = df_salary.apply(project_salary,axis = 1)
```

```python
df_salary[['job_title_short','salary_year_avg','inflated_salary']]
```

```python
df_salary['inflated_salary_lambda'] = df_salary.apply(lambda row: 1.05 * row['salary_year_avg'] if 'Senior' in row['job_title_short'] else row['salary_year_avg'] * 1.03, axis = 1)
```
